```python
# comparative cohort visualization

import pandas as pd
import matplotlib.pyplot as plt

comparison = pd.DataFrame({
    'Metformin': metformin_group['readmitted'].value_counts(normalize=True),
    'Insulin': insulin_group['readmitted'].value_counts(normalize=True)
})

comparison.plot(kind='bar')

plt.title("Readmission Comparison")
plt.ylabel("Percentage")
plt.show()
```

## Initial Findings

- Metformin and insulin cohorts show different readmission distributions.
- Insulin patients appear to have longer hospital stays.
- Medication burden may be higher among insulin users.

```python
# Import Statistics Library

from scipy.stats import chi2_contingency
from scipy.stats import ttest_ind
```

```python
# Create Readmission Contingency Table

readmission_table = pd.crosstab(
    df['metformin'] != 'No',
    df['readmitted']
)

readmission_table
```

```python
# Run Chi-Square Test

chi2, p, dof, expected = chi2_contingency(readmission_table)

print("P-value:", p)
```

## Chi-Square Test Result

The readmission distribution differed significantly between medication groups (p < 0.05), suggesting a potential association between medication exposure and hospital readmission outcomes.

```python
# Compare Hospital Stay Statistically

ttest_ind(
    metformin_group['time_in_hospital'],
    insulin_group['time_in_hospital'],
    nan_policy='omit'
)
```

## T-Test Result

Average hospital stay differed significantly between medication cohorts, indicating potential differences in clinical severity or treatment outcomes.

```python


import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:/Users/dharm/Project_2_RWE_Drug_Effectiveness/data/diabetic_data.csv")

metformin_group = df[df['metformin'].isin(['Steady', 'Up', 'Down'])]

insulin_group = df[df['insulin'].isin(['Steady', 'Up', 'Down'])]
```

```python
# Create Readmission Chart

comparison = pd.DataFrame({
    'Metformin': metformin_group['readmitted'].value_counts(normalize=True),
    'Insulin': insulin_group['readmitted'].value_counts(normalize=True)
})

comparison.plot(kind='bar', figsize=(8,5))

plt.title("Readmission Rate Comparison")
plt.ylabel("Percentage")
plt.xlabel("Readmission Category")

plt.xticks(rotation=0)


plt.show()
```

```python
# Create Hospital Stay Boxplot

plot_data = [
    metformin_group['time_in_hospital'],
    insulin_group['time_in_hospital']
]

plt.figure(figsize=(8,5))

plt.boxplot(plot_data, labels=['Metformin', 'Insulin'])

plt.title("Hospital Stay Distribution")
plt.ylabel("Days in Hospital")

plt.show()
```

```python
# Compare Average Metrics

summary = pd.DataFrame({
    'Metric': [
        'Avg Hospital Stay',
        'Avg Medications'
    ],
    
    'Metformin': [
        metformin_group['time_in_hospital'].mean(),
        metformin_group['num_medications'].mean()
    ],
    
    'Insulin': [
        insulin_group['time_in_hospital'].mean(),
        insulin_group['num_medications'].mean()
    ]
})

summary
```
